```python
# ==========================================
# Interactive Clickbait & Fake News Detector
# Google Colab Copy-Paste Ready Script
# ==========================================

import subprocess
import sys
import pandas as pd

def install_package(package):
    try:
        __import__(package)
    except ImportError:
        subprocess.check_call([sys.executable, "-m", "pip", "install", package])

print("--- Checking and installing required libraries ---")
install_package("pandas")
install_package("scikit-learn")

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

# 1. Training Dataset (Headlines labeled as Reliable or Clickbait)
data = {
    'headline': [
        "Local council approves new community park funding for next year.",
        "You Won't Believe What This Celebrity Did In Public! Shocking!",
        "Quarterly financial report shows steady growth across tech sector.",
        "Doctors Hate Her! Discover This One Weird Trick to Lose 10 Pounds.",
        "Scientists discover new exoplanet with potential water signs.",
        "Top 10 Reasons Why Your Morning Coffee Is Ruining Your Life.",
        "Global climate summit reaches consensus on carbon emission goals.",
        "Click Here to Claim Your Free iPhone Right Now Before It's Gone!"
    ],
    'label': [
        'Reliable', 'Clickbait',
        'Reliable', 'Clickbait',
        'Reliable', 'Clickbait',
        'Reliable', 'Clickbait'
    ]
}

df = pd.DataFrame(data)

print("\n--- Training Machine Learning Classification Model ---")
# Build and train classification pipeline
model = Pipeline([
    ('tfidf', TfidfVectorizer(stop_words='english')),
    ('clf', LogisticRegression())
])
model.fit(df['headline'], df['label'])

# 2. Interactive Loop for Google Colab
print("\n" + "="*60)
print("🚨 INTERACTIVE CLICKBAIT DETECTOR READY")
print("Type a news headline below to test it.")
print("Type 'exit' or 'quit' to stop.")
print("="*60 + "\n")

while True:
    try:
        user_headline = input("Enter headline: ").strip()
    except EOFError:
        break

    if user_headline.lower() in ['exit', 'quit']:
        print("\nExiting detector. Have a great day!")
        break

    if not user_headline:
        print("⚠️ Please enter a valid headline.\n")
        continue

    # Predict label
    prediction = model.predict([user_headline])[0]

    print(f"\n[AI Analysis]")
    print(f"👉 Classification: {prediction}")
    print("-" * 60 + "\n")
```
